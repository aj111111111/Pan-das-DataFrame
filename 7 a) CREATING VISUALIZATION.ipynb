```python
df.plot.scatter(x = 'Texture Rating', y = 'Overall Rating', s = 500, c = 'Yellow')
```

df.plot.scatter(...): Tells Pandas to create a scatter plot using your DataFrame (df), where each row in your dataset is represented as a single dot on the graph.

x='Texture Rating': Places the Texture Rating values along the horizontal X-axis.

y='Overall Rating': Places the Overall Rating values along the vertical Y-axis. Every dot's position tells you an item's specific combination of texture and overall score.

s=500: Sets the size (s) of every marker to 500. This is significantly larger than the default size, making the points bold and prominent (sometimes called a bubble chart style).

c='Yellow': Paints every data point yellow.

How to read the resulting plot:
Correlation: Look at how the yellow dots trend. If the dots slope upward from bottom-left to top-right, it means items with better texture tend to have higher overall ratings. If they are scattered randomly or slope downward, texture has little or no positive impact on the overall score.

Clustering: Notice where the dots pack tightly together. That concentration highlights the most common "Texture vs. Overall Rating" profile across your dataset.

A boxplot (box-and-whisker plot) provides a visual summary of the distribution, central tendency, variability, and potential outliers of numerical columns in your DataFrame.

Anatomy of the Box

Median (Q2): The horizontal line inside the box representing the 50th percentile (the exact middle of the data).

Bottom of the Box (Q1): The 25th percentile (lower quartile). 25% of the data falls below this value.

Top of the Box (Q3): The 75th percentile (upper quartile). 75% of the data falls below this value.

The Box (IQR): The height of the box represents the Interquartile Range ($IQR = Q3 - Q1$), containing the middle 50% of your data.

Whiskers and Outliers

Whiskers: Lines extending outward from the top and bottom of the box to indicate the spread of the rest of the dataset, typically stretching up to $1.5 \times \text{IQR}$ from the quartiles.

Outliers: Individual dots, circles, or asterisks plotted past the ends of the whiskers, flagging unusually high or low data points that warrant inspection.

Key Interpretations

Skewness: If the median line sits closer to the bottom of the box, the data is right-skewed (positively skewed, with a long tail of high values). If it sits closer to the top, it is left-skewed.

Spread and Variability: A taller box and longer whiskers mean your data is widely dispersed. A short, compact box indicates values are tightly clustered around the median.

Side-by-Side Comparisons: When df.boxplot() generates multiple plots for different columns, you can instantly compare their medians and ranges to spot which columns have higher values or greater variance.
