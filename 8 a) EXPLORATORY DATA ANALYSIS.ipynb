# EXPLORATORY DATA ANALYSIS (EDA)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"world_population.csv")
df
```

```python
pd.set_option('display.float_format', lambda x: '%.2f' % x) #upto 2 decimal
```

```python
df.info()
```

```python
df.describe()      # will give values like mean. min,max,count, std
```

```python
df.isnull().sum()   # total null or empty cell in each columns
```

```python
df.nunique()     # unique values in each columns
```

```python
df.sort_values(by="World Population Percentage", ascending=False).head(10)
```

```python
df.corr(numeric_only=True)
```

```python
sns.heatmap(df.corr(numeric_only=True), annot = True)

plt.rcParams['figure.figsize'] = (20,7)

plt.show()
```

```python
df.groupby('Continent').mean(numeric_only=True).sort_values(by="2022 Population",ascending=False)
# grouping, mean, sorting 
```

```python
df[df['Continent'].str.contains('Oceania')]  # filtering out 
```

```python
df2 = df.groupby('Continent')[['1970 Population',
       '1980 Population', '1990 Population', '2000 Population',
       '2010 Population', '2015 Population', '2020 Population',
       '2022 Population']].mean().sort_values(by="2022 Population",ascending=False)
df2
```

```python
df.columns
```

```python
df3 = df2.transpose()
df3
```
